```python
class Employee:
    language = "Python"    
    salary = 50000          

    def getInfo(self):
        print(f"Language is {self.language}.  Salary is {self.salary}")

    def greet(self):
        print("Hello, welcome to the company!")

sam = Employee()
sam.language = "JavaScript"      # instance
sam.greet()
sam.getInfo()
# Employee.getInfo(sam)      # class
```

```python
class Employee:
    language = "Python"    
    salary = 50000          

    def getInfo(self):
        print(f"Language is {self.language}.  Salary is {self.salary}")

    @staticmethod       # static method
    
    def greet():
        print("Hello, welcome to the company!")

sam = Employee()
sam.language = "JavaScript"      # instance
sam.greet()
sam.getInfo()
# Employee.getInfo(sam)      # class
```
