1. Create a class “Programmer” for storing information of few programmers working at
Microsoft.

```python
class Programmer:
    compony = "Microsoft"
    def __init__(self, name, salary, pin):
        self.name = name
        self.salary = salary
        self.pin = pin

p = Programmer("John", 60000, 467869)
print(p.name, p.salary, p.pin, p.compony)

p2 = Programmer("Sam", 50000, 132789)
print(p2.name, p2.salary, p2.pin, p2.compony)

p3 = Programmer("Alice", 70000, 648390)
print(p3.name, p3.salary, p3.pin, p3.compony)
```

2. Write a class “Calculator” capable of finding square, cube and square root of a number.

```python
class Calculator:
    def __init__(self, n):
        self.n = n

    def square(self):
        print(f"The square of {self.n} is {self.n * self.n}")

    def cube(self):
            print(f"The cube of {self.n} is {self.n * self.n * self.n}")

    def squareroot(self):
            print(f"The square root of {self.n} is {self.n ** 1/2}")

a = Calculator(4)
a.square()
a.cube()
a.squareroot()

```

3. Create a class with a class attribute a; create an object from it and set ‘a’ directly using
‘object.a = 0’. Does this change the class attribute?

```python
class Sample:
    a = 5

o = Sample()
print(o.a)  # instance variable

o.a = 10
print(o.a)  # instance variable

print(Sample.a)  # class variable
```

4. Add a static method in problem 2, to greet the user with hello.

```python
class Calculator:
    def __init__(self, n):
        self.n = n

    def square(self):
        print(f"The square of {self.n} is {self.n * self.n}")

    def cube(self):
            print(f"The cube of {self.n} is {self.n * self.n * self.n}")

    def squareroot(self):
            print(f"The square root of {self.n} is {self.n ** 1/2}")

    @staticmethod
    def greet():
        print("Hello, welcome to the calculator!")

a = Calculator(4)
a.greet()
a.square()
a.cube()
a.squareroot()

```

5. Write a Class ‘Train’ which has methods to book a ticket, get status (no of seats) and get
fare information of train running under Indian Railways.

```python
from random import randint

class Train:

    def __init__(self, trainNo):
        self.trainNo = trainNo

    def book(self, fro, to):
        print(f"Train booked for {self.trainNo} from {fro} to {to}")

    def getStatus(self):
        print(f"Train no : {self.trainNo} is on time")

    def getFare(self, fro, to):
        print(f"Ticket is fare in train no : {self.trainNo} from {fro} to {to} is {randint(100, 1000)}")

train = Train(12345)
train.book("Delhi", "Mumbai")
train.getStatus()
train.getFare("Delhi", "Mumbai")
```
