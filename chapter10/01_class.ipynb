```python
class Employee:
    name = "John"
    language = "Python"
    salary = 50000

john = Employee()
print(john.name, john.language, john.salary)
```

```python
class Employee:
    language = "Python"     # class attribute
    salary = 50000          # class attribute

john = Employee()
print(john.language, john.salary)

sam = Employee()
sam.name = "Sam"      # object attribute
print(sam.name, sam.language, sam.salary)
```
