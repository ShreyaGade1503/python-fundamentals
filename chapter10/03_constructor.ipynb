```python
class Employee:
    language = "Python"    
    salary = 50000          

    def __init__(self , name, salary, language):  # dunder method which is called when an object is created
        self.name = name
        self.salary = salary
        self.language = language
        print("object")

    def getInfo(self):
        print(f"Language is {self.language}.  Salary is {self.salary}")

    @staticmethod       # static method
    
    def greet():
        print("Hello, welcome to the company!")

sam = Employee("John", 60000, "JavaScript")      
#sam.name = "Sam"                
print(sam.name, sam.salary, sam.language)
```
